```python
import pandas as pd

df = pd.read_csv("cleaned_employee_dataset.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
# Basic statistical summary

print(df.describe())
```

```python
import matplotlib.pyplot as plt

# Age distribution
plt.figure(figsize=(8, 5))

plt.hist(df["Age"], bins=8, edgecolor="black")

plt.title("Age Distribution of Employees")
plt.xlabel("Age")
plt.ylabel("Number of Employees")

plt.show()
```

```python
# Performance score distribution

performance_counts = df["Performance_Score"].value_counts()

plt.figure(figsize=(8, 5))

plt.bar(performance_counts.index, performance_counts.values, edgecolor="black")

plt.title("Employee Performance Score Distribution")
plt.xlabel("Performance Score")
plt.ylabel("Number of Employees")

plt.show()
```

```python
# Relationship between Age and Salary

plt.figure(figsize=(8, 5))

plt.scatter(df["Age"], df["Salary"])

plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")

plt.show()
```

```python
# Check salary distribution and outliers

plt.figure(figsize=(8, 4))

plt.boxplot(df["Salary"])

plt.title("Salary Distribution and Outliers")
plt.ylabel("Salary")

plt.show()
```

Task 2: Exploratory Data Analysis (EDA)

The cleaned employee dataset was analyzed using Pandas and Matplotlib.

Created a histogram to understand the age distribution.
Created a bar chart to compare employee performance scores.
Used a scatter plot to analyze the relationship between age and salary.
Used a box plot to identify salary outliers.
The analysis showed that employees aged 30 had the highest concentration.
Performance categories were fairly balanced, while Poor performance was comparatively lower.
No clear linear relationship was observed between age and salary.
No significant salary outliers were visible in the box plot.

Conclusion: The EDA helped identify important patterns and understand the distribution and relationships within the employee dataset.
