```python
import pandas as pd

df = pd.read_csv('/content/Messy_Employee_dataset.csv')

df.head()
```

```python
print("Dataset Shape:", df.shape)

df.head()
```

```python
# Check dataset information
df.info()
```

```python
# Check missing values in each column

missing_values = df.isnull().sum()

print(missing_values)
```

```python
# Check age statistics

print(df["Age"].describe())
```

```python
# Check age value distribution

print(df["Age"].value_counts().sort_index())
```

```python
# Fill missing Age values with the median

age_median = df["Age"].median()

df["Age"] = df["Age"].fillna(age_median)

print("Median Age used:", age_median)
print("Missing Age values after cleaning:", df["Age"].isnull().sum())
```

```python
# Check salary statistics

print(df["Salary"].describe())
```

```python
# Fill missing Salary values with the median

salary_median = df["Salary"].median()

df["Salary"] = df["Salary"].fillna(salary_median)

print("Median Salary used:", salary_median)
print("Missing Salary values after cleaning:", df["Salary"].isnull().sum())
```

```python
# Final missing values check

print(df.isnull().sum())
```

```python
# Check for duplicate rows

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Check current data type of Join_Date

print("Before conversion:", df["Join_Date"].dtype)
```

```python
# Convert Join_Date from object to datetime

df["Join_Date"] = pd.to_datetime(df["Join_Date"])

print("After conversion:", df["Join_Date"].dtype)
```

```python
# Check data types after cleaning

print(df.dtypes)
```

```python
# Check phone number values

print(df["Phone"].head(10))
print("\nData type:", df["Phone"].dtype)
```

```python
# Check phone number range

print("Minimum Phone value:", df["Phone"].min())
print("Maximum Phone value:", df["Phone"].max())

print("\nNumber of negative phone values:",
      (df["Phone"] < 0).sum())
```

```python
# Check whether negative phone values follow any pattern

print(df["Phone"].describe())
```

```python
# Check if Phone column contains any valid positive values

valid_phone_count = (df["Phone"] > 0).sum()

print("Valid positive phone values:", valid_phone_count)
print("Invalid negative phone values:", (df["Phone"] < 0).sum())
```

```python
# Remove the invalid Phone column

df = df.drop(columns=["Phone"])

print("Phone column removed.")
print("New dataset shape:", df.shape)
```

```python
# Display basic statistics for numerical columns

print(df.describe())
```

```python
# Count employees by department and region

print(df["Department_Region"].value_counts())
```

```python
# Count employees by status

print(df["Status"].value_counts())
```

```python
# Count employees by performance score

print(df["Performance_Score"].value_counts())
```

```python
# Final dataset verification

print("Final dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```

```python
# Save the cleaned dataset as a CSV file

df.to_csv("cleaned_employee_dataset.csv", index=False)

print("Cleaned dataset saved successfully!")
```
