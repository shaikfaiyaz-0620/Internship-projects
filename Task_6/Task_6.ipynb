```python
import pandas as pd

df6 = pd.read_csv("Superstore Sales Dataset.csv")

print(df6.shape)
```

```python
print("Total Sales:", round(df6["Sales"].sum(), 2))
print("Average Sales:", round(df6["Sales"].mean(), 2))
```

```python
category_sales = df6.groupby("Category")["Sales"].sum().sort_values(ascending=False)

print(category_sales)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.bar(category_sales.index, category_sales.values)

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```

```python
region_sales = df6.groupby("Region")["Sales"].sum().sort_values(ascending=False)

print(region_sales)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(region_sales.index, region_sales.values)

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")

plt.show()
```

```python
subcategory_sales = df6.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False)

print(subcategory_sales)
```

```python
top10_subcategories = subcategory_sales.head(10)

plt.figure(figsize=(10, 6))

plt.barh(
    top10_subcategories.index[::-1],
    top10_subcategories.values[::-1]
)

plt.title("Top 10 Sub-Categories by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Sub-Category")

plt.show()
```

```python
df6["Order Date"] = pd.to_datetime(
    df6["Order Date"],
    format="%d/%m/%Y"
)

yearly_sales = df6.groupby(
    df6["Order Date"].dt.year
)["Sales"].sum()

print(yearly_sales)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    yearly_sales.index,
    yearly_sales.values,
    marker="o"
)

plt.title("Yearly Sales Trend")
plt.xlabel("Year")
plt.ylabel("Total Sales")
plt.grid(True)

plt.show()
```

Final Business Insights
Technology was the highest-selling category with total sales of approximately 827,455.87.
West was the best-performing region with sales of approximately 710,219.68, while South had the lowest sales.
Among sub-categories, Phones generated the highest sales, followed closely by Chairs.
Annual sales showed an overall growth trend, increasing from approximately 479,856 in 2015 to 722,052 in 2018.
The analysis shows that Technology products and the West region are strong contributors to overall sales.
These insights can help businesses focus on high-performing categories, products, and regions to improve sales performance.
